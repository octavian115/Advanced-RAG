## Recursive loader

```python
# scraps even the child link
# needs just the base loader
```

```python
from langchain_community.document_loaders import RecursiveUrlLoader
```

```python
base_url = "https://docs.langchain.com/oss/python/integrations/document_loaders"

loader = RecursiveUrlLoader(url=base_url,
                            max_depth=2)
```

```python
docs = loader.load()
```

```python
len(docs)
```

```python
# takes too much time
# loads everything in memory
# so we use lazy load
# returns a generator object
# loads item by item
# then deletes it
```

```python
# lazy load
documents_lazy_load = loader.lazy_load()
```

```python
documents_lazy_load
```

```python
counter = 0
for doc in documents_lazy_load:
    # stop condition
    if counter == 20:
        break

    counter+= 1

    pp(doc.page_content[0:300])
    pp(doc.metadata)
```

```python
# we don't need to wait for entire process
# start getting results as soon as possible
# saved memory and time
```
